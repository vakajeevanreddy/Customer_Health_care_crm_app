```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Create a sample plot so plt.show() has something to render
plt.figure(figsize=(8, 4))
sns.histplot(df['Age'], kde=True, color='teal')
plt.title('Age Distribution')

# Now display it
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(df['Age'], kde=True, bins=30, color='teal')
plt.title('Distribution of Patient Age')

plt.subplot(1, 2, 2)
sns.histplot(df['Billing Amount'], kde=True, bins=30, color='purple')
plt.title('Distribution of Billing Amount')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14,4))
plt.subplot(1,3,1)
sns.countplot(data = df, x = "Medical Condition" , palette="Set2")
plt.subplot(1,3,2)
sns.countplot(data=df, x = "Admission Type" , palette="Set1")
plt.subplot(1,3,3)
sns.countplot(data=df, x = "Test Results" , palette="Pastel1")
plt.title("Test Results Count")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Medical Condition', y='Billing Amount', palette='coolwarm')
plt.title('Billing Amount across Different Medical Conditions')
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
# Select only numerical columns for correlation
numeric_df = df.select_dtypes(include=['float64', 'int64'])
correlation_matrix = numeric_df.corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.violinplot(data =  df , x = "Insurance Provider" , y = "Billing Amount" , palette="muted")
plt.title('Billing Amount Distribution by Insurance Provider')
plt.xlabel('Insurance Provider')
plt.ylabel('Billing Amount')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(6,6))
admission_counts = df["Admission Type"].value_counts()
plt.pie(admission_counts, labels= admission_counts.index, autopct="%1.2f%%", colors=["red", "blue", "black"],
        startangle=90)
plt.title("Proportion of Admission Types")
plt.show()
```
